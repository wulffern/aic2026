# Maxwell's equations 

There are equations that relate electric field, magnetic field, charge density and current density to each-other. 

Electric flux is the total electric field through a surface. Magnetic flux is the total magnetic field through a surface. 

$$ \oint_{\partial \Omega} \mathbf{E} \cdot d\mathbf{S} = \frac{1}{\epsilon_0} \iiint_{V} \rho
\cdot dV$$  

,relates net electric flux to net enclosed electric charge


$$ \oint_{\partial \Omega} \mathbf{B} \cdot d\mathbf{S} = 0$$

,relates net magnetic flux to net enclosed magnetic charge


$$ \oint_{\partial \Sigma} \mathbf{E} \cdot d\mathbf{\ell} = - \frac{d}{dt}\iint_\Sigma \mathbf{B}
\cdot d\mathbf{S}$$

,relates induced electric field to changing magnetic flux


$$ \oint_{\partial \Sigma} \mathbf{B} \cdot d\mathbf{\ell} = \mu_0\left(
\iint_\Sigma \mathbf{J} \cdot d\mathbf{S} + \epsilon_0 \frac{d}{dt}\iint_\Sigma
\mathbf{E} \cdot d\mathbf{S} \right)$$

,relates induced magnetic field to changing electric flux and to current

## Ignore displacement current

When can we ignore displacement current? Let's figure out. 

```python
import scipy.constants as const
epsilon0 = const.epsilon_0

Rho = 1.68e-8 # Resistance of copper
w = 254e-6    # width of PCB track
h = 35e-6     # thickness of PCB track
L = 10e-3     # Length of loop
I = 100e-3    # Current step in loop
dt = 1e-12     # Time for current change

R = Rho*L/(w*h)
print(f"R={R:.3} Ohm")

# Electric field in loop 
V = I*R
E = V/L
print(f"E={E:.3} V/m")

#Electric flux in loop 
flux = E*w*h
print(f"flux={flux:.3} V*m")

#Notice that the flux is independent of w, h, and L, and proportional to I and Rho 
flux1 = I*Rho
print(f"flux1={flux1:.3} V*m")

#Displacement current in conductor (Ignore the fact that the permittivity is complex in copper)
Id = epsilon0*flux/dt
print(f"Id={Id:.3} A")

# Current in loop 
Itotal = I + Id
print(f"Itotal={Itotal:.4} A")
```
